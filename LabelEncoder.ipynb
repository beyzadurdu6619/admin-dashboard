```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('tour.csv')
df.head()
#https://www.kaggle.com/datasets/gulliverwoods/tour-de-france-winner-data?resource=download
```

```python
le=LabelEncoder()
le.fit(df["Team"])
```

```python
x=le.transform(df["Team"])
```

```python
x
```

```python
le.inverse_transform(x)
```
